```python
import pandas as pd
import matplotlib.pyplot as plt

players = pd.read_csv('../data/processed/player_stats.csv')

# Top 10 Run Scorers
top_runs = players.nlargest(10, 'runs')

plt.bar(top_runs['player'], top_runs['runs'])
plt.xticks(rotation=45)
plt.title('Top 10 Run Scorers')
plt.savefig('../visualizations/top_run_scorers.png')
plt.show()
```
